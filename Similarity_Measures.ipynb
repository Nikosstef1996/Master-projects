```python
from collections import Counter
def cosine_similarity(doc1, doc2):
    # TODO: Implement cosine similarity from scratch
    # You may build a term frequency vector for each document
    counts_doc1 = Counter(doc1)
    l1 = len(doc1)
    counts_doc2 = Counter(doc2)
    l2 = len(doc2)
    for i in counts_doc1:
      counts_doc1[i] /= l1
    for i in counts_doc2:
      counts_doc2[i] /= l2
    all_words = set(counts_doc1.keys()) | set(counts_doc2.keys())
    mul_i = 0
    array1_sq = 0
    array2_sq = 0
    for i in all_words:
      mul_i += counts_doc1[i] * counts_doc2[i]
      array1_sq += counts_doc1[i] * counts_doc1[i]
      array2_sq += counts_doc2[i] * counts_doc2[i]

    return mul_i / (sqrt(array1_sq) * sqrt(array2_sq))
    pass
```

```python
def jaccard_similarity(doc1, doc2):
    # TODO: Implement Jaccard similarity from scratch
    intersection_set_size = 0
    union_set_size = 0
    for i, j in zip(doc1, doc2):
        if i==j:
            intersection_set_size += 1
            union_set_size += 1
        else:
            union_set_size += 2
    return intersection_set_size / union_set_size
    pass
```

```python
from IPython.core.interactiveshell import dis
def manhattan_similarity(doc1, doc2):
    # TODO: Implement Manhattan similarity from scratch
    c1 = Counter(doc1)
    c2 = Counter(doc2)
    all_words = set(c1.keys()) | set(c2.keys())
    distance = sum(abs(c1[word] - c2[word]) for word in all_words)
    return 1 - distance/len(all_words)
    pass
```

```python
def euclidean_similarity(doc1, doc2):
    # TODO: Implement Euclidean similarity from scratch
    c1 = Counter(doc1)
    c2 = Counter(doc2)
    all_words = set(c1.keys()) | set(c2.keys())
    distance = sum(abs(c1[word] - c2[word])**2 for word in all_words)
    return 1 - distance/len(all_words)
    pass
```

```python
# We implement TF-IDF and use it to compute cosine similarity

def cosine_similarity_tf_idf(docs, doc1, doc2):

  # First we compute tf for a document. It returns a dictionary of the keys of a document with the frequencies of the words appearing.
  def compute_tf(document):
    tf_doc = Counter(document)
    l = len(document)
    for i in tf_doc:
      tf_doc[i] /= l
    return tf_doc

  # Now we compute idf for the list of documents for each term. It returns a dictionary with the idf for each term.
  def compute_idf(docs):
    l = len(docs)
    unique_words = set(s for sublist in docs for s in sublist)
    num_docs = {}
    idf = {}

    for word in unique_words:
        count_in_docs = 0
        for document in docs:
            if word in document:
                count_in_docs += 1
        num_docs[word] = count_in_docs

        if num_docs[word] > 0:
            idf[word] = math.log(l / num_docs[word])
        else:
            idf[word] = 0
    return idf

  # Now we compute tf-idf for each of the two documents:
  unique_words = set(s for sublist in docs for s in sublist)
  tf_idf1={}
  tf_idf2={}
  for word in unique_words:
    tf_idf1[word] = compute_tf(doc1)[word] * compute_idf(docs)[word]
    tf_idf2[word] = compute_tf(doc2)[word] * compute_idf(docs)[word]

  # We compute here the inner product :
  def inner_product(dic1, dic2, keys):
    mul_i = 0
    array1_sq = 0
    array2_sq = 0
    for word in keys:
      mul_i += dic1[word] * dic2[word]
      array1_sq += dic1[word] * dic1[word]
      array2_sq += dic2[word] * dic2[word]

    if array1_sq == 0 or array2_sq == 0:
      return 0
    else:
      return mul_i / (sqrt(array1_sq) * sqrt(array2_sq))


  return inner_product(tf_idf1, tf_idf2, unique_words)
  pass



```
